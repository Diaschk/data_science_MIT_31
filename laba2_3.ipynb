```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd


np.random.seed(42) 

m = 3
b = -4
x = np.arange(-5, 6, 1) 

y_true = m * x + b
error = np.random.normal(0, 2, size=x.shape) 
y_measured = y_true + error



plt.figure(figsize=(8,5))
plt.plot(x, y_true, label=" y=3x-4", color="black")
plt.scatter(x, y_measured, color="red", label="Виміряні дані з похибкою")
plt.title("Лінійна функція y = 3x - 4 з похибкою")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()



mae = np.mean(np.abs(y_measured - y_true))
mse = np.mean((y_measured - y_true)**2)

print(f"MAE: {mae:.3f}")
print(f"MSE: {mse:.3f}")



dataframe_results = pd.DataFrame({
    "x": x,
    "y_true": y_true,
    "y_measured": y_measured,
    "error": y_measured - y_true
})

dataframe_results.to_csv("laba2_3.csv", index=False)
print("Дані збережено у файл laba2_3.csv")
```
