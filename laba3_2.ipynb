```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python

df_kaggle = pd.read_csv('car_prediction_data.csv')

print(df_kaggle.head())
```

```python

X_kaggle = df_kaggle[['Present_Price']]  
y_kaggle = df_kaggle['Selling_Price']    


model_kaggle = LinearRegression()
model_kaggle.fit(X_kaggle, y_kaggle)


y_pred_kaggle = model_kaggle.predict(X_kaggle)


mse_k = mean_squared_error(y_kaggle, y_pred_kaggle)
mae_k = mean_absolute_error(y_kaggle, y_pred_kaggle)
r2_k = r2_score(y_kaggle, y_pred_kaggle)


print(f"MSE: {mse_k:.2f}")
print(f"MAE: {mae_k:.2f}")
print(f"R^2: {r2_k:.2f}")
```

```python


plt.scatter(X_kaggle, y_kaggle, color='red', label='Фактичні дані')
plt.plot(X_kaggle, y_pred_kaggle, color='black', linewidth=2, label='Лінія регресії')
plt.xlabel('Present_Price')
plt.ylabel('Selling_Price')
plt.title('Лінійна регресія — Car Price Prediction')
plt.legend()
plt.show()


```
