```python
import tensorflow as tf
import matplotlib.pyplot as plt


scalar = tf.constant(20)
vector = tf.constant([5, 6, 7])
matrix = tf.constant([[1, 2], [3, 4]])
tensor = tf.constant([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])


print(f"{'Type':<10} {'Форма':<15} {'Ранг':<5} {'Розмір':<5}")
print(f"{'Scalar':<10} {str(scalar.shape):<15} {tf.rank(scalar).numpy():<5} {tf.size(scalar).numpy():<5}")
print(f"{'Vector':<10} {str(vector.shape):<15} {tf.rank(vector).numpy():<5} {tf.size(vector).numpy():<5}")
print(f"{'Matrix':<10} {str(matrix.shape):<15} {tf.rank(matrix).numpy():<5} {tf.size(matrix).numpy():<5}")
print(f"{'Tensor':<10} {str(tensor.shape):<15} {tf.rank(tensor).numpy():<5} {tf.size(tensor).numpy():<5}")


t1 = tf.random.uniform([5, 300], 0, 1)
t2 = tf.random.uniform([5, 300], 0, 1)


matmul_result = tf.matmul(t1, tf.transpose(t2))
print("\nМноження матриць:", matmul_result.shape)


dot_result = tf.tensordot(t1, tf.transpose(t2), axes=1)
print("\nСкалярний добуток:", dot_result.shape)



image_tensor = tf.random.uniform([224, 224, 3], 0, 1)
print("\nТензор з випадковими значеннями:", image_tensor.shape)



print("\nМінімальне значення:", tf.reduce_min(image_tensor).numpy(), "Максимальне значення:", tf.reduce_max(image_tensor).numpy())


tensor_4d = tf.random.uniform([1, 224, 224, 3])
tensor_squeezed = tf.squeeze(tensor_4d)
print("\nЗмінення форми тензора:", tensor_squeezed.shape)


values = tf.constant([9, 5, 3, 4, 1, 8, 9, 2, 6, 7])
max_index = tf.argmax(values)
print("\nІндекс, який має максимальне значення:", max_index.numpy())


one_hot = tf.one_hot(max_index, depth=10)
print("\nЗакодований тензор:", one_hot.numpy())


print("\nGPU:", len(tf.config.list_physical_devices('GPU')))




X_reg = tf.random.uniform([100], -5, 5)
y_reg = 2 * X_reg + 3 + tf.random.normal([100], 0, 1)
plt.scatter(X_reg, y_reg)
plt.title("Набір даних для регресії")
plt.xlabel("X")
plt.ylabel("y")
plt.show()


X_class0 = tf.random.normal([50, 2], mean=-2, stddev=1)
X_class1 = tf.random.normal([50, 2], mean=2, stddev=1)
X_class = tf.concat([X_class0, X_class1], axis=0)
y_class = tf.concat([tf.zeros(50), tf.ones(50)], axis=0)
plt.scatter(X_class[:, 0], X_class[:, 1], c=y_class, cmap="coolwarm")
plt.title("Набір даних для класифікації")
plt.xlabel("X1")
plt.ylabel("X2")
plt.show()
```
