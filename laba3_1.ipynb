Прогнозування кількості відвідувачів кафе залежно від температури повітря

```python
import pandas as pd
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
X = np.random.uniform(0, 35, 100).reshape(-1, 1)  
y = 20 + 5 * X.flatten() + np.random.normal(0, 10, 100)


df = pd.DataFrame({'Temperature': X.flatten(), 'Visitors': y})
print(df.head())
```

```python

X_train, X_test, y_train, y_test = train_test_split(df[['Temperature']], df['Visitors'], test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)


y_pred = model.predict(X_test)


mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
print(f'MSE: {mse:.2f}, MAE: {mae:.2f}')
```

```python


plt.figure(figsize=(6,4))
plt.scatter(X_test, y_test, color='red', label='Реальні дані')
plt.plot(X_test, y_pred, color='black', label='Прогноз')
plt.title('Прогноз відвідувачів кафе залежно від температури')
plt.xlabel('Температура (°C)')
plt.ylabel('Кількість відвідувачів')
plt.legend()
plt.grid(True)
plt.show()

```
