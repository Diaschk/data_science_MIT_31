```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
import numpy as np
import os
import zipfile
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
from PIL import Image
import requests
from io import BytesIO

print("Підготовка середовища")
print(f"TensorFlow версія: {tf.__version__}")


physical_devices = tf.config.experimental.list_physical_devices('GPU')
if physical_devices:
    tf.config.experimental.set_memory_growth(physical_devices[0], True)
    print("GPU активовано")

print("\nСтворення набору даних: орхідеї та троянди")

def create_dataset_structure():
    base_dir = "plants_dataset"
    folders = [
        f"{base_dir}/train/orchids",
        f"{base_dir}/train/roses", 
        f"{base_dir}/test/orchids",
        f"{base_dir}/test/roses"
    ]
    
    for folder in folders:
        os.makedirs(folder, exist_ok=True)
    
    return base_dir

base_dir = create_dataset_structure()
print("Структура папок створена")

print("\nЗавантаження зображень орхідей та троянд")

def download_plants_images():
    print("Створення демо-датасету орхідей та троянд")
    
    orchid_colors = [
        (255, 192, 203),  
        (238, 130, 238), 
        (221, 160, 221),  
        (255, 105, 180),  
    ]
    
    rose_colors = [
        (255, 0, 0),     
        (255, 20, 147),  
        (199, 21, 133),   
        (255, 99, 71),   
    ]
    
    def create_flower_image(flower_type, index):
        img = np.random.randint(100, 200, (224, 224, 3), dtype=np.uint8)  
        
        if flower_type == 'orchids':
            colors = orchid_colors
            center_x, center_y = 112, 112
            for i in range(5):  
                angle = 2 * np.pi * i / 5
                radius = 40 + np.random.randint(-10, 10)
                x = int(center_x + radius * np.cos(angle))
                y = int(center_y + radius * np.sin(angle))
                
                for dx in range(-25, 26):
                    for dy in range(-15, 16):
                        if dx*dx/625 + dy*dy/225 <= 1:  
                            nx, ny = x + dx, y + dy
                            if 0 <= nx < 224 and 0 <= ny < 224:
                                color_idx = (index + i) % len(colors)
                                img[ny, nx] = colors[color_idx]
                                
        else: 
            colors = rose_colors
            center_x, center_y = 112, 112
            for layer in range(3):  
                radius = 30 - layer * 8
                color_idx = (index + layer) % len(colors)
                color = colors[color_idx]
                
                for angle in np.linspace(0, 2*np.pi, 100):
                    for r in range(radius, radius + 15):
                        x = int(center_x + r * np.cos(angle))
                        y = int(center_y + r * np.sin(angle))
                        if 0 <= x < 224 and 0 <= y < 224:
                            if np.random.random() > 0.3:
                                img[y, x] = color
        
        return Image.fromarray(img)
    
    for class_name in ['orchids', 'roses']:
        print(f"Створення {class_name}")
        
        for i in range(80):  
            img = create_flower_image(class_name, i)
            img.save(f"{base_dir}/train/{class_name}/img_{i:03d}.jpg")
        
        for i in range(20):  
            img = create_flower_image(class_name, i + 1000)  
            img.save(f"{base_dir}/test/{class_name}/img_{i:03d}.jpg")
    
    print("Датасет орхідей та троянд створено\n")

download_plants_images()


IMG_SIZE = (128, 128)
BATCH_SIZE = 32


train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=30, 
    width_shift_range=0.3,
    height_shift_range=0.3,
    horizontal_flip=True,
    vertical_flip=True,  
    zoom_range=0.3,
    shear_range=0.2,
    brightness_range=[0.7, 1.3],  
    channel_shift_range=0.2, 
    fill_mode='nearest'
)


test_datagen = ImageDataGenerator(rescale=1./255)


train_data = train_datagen.flow_from_directory(
    f'{base_dir}/train',
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='binary',
    classes=['orchids', 'roses'],  
    seed=42
)

test_data = test_datagen.flow_from_directory(
    f'{base_dir}/test',
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='binary',
    classes=['orchids', 'roses'],
    seed=42,
    shuffle=False
)

print(f"\nІнформація про дані:")
print(f"   Тренувальні зображення: {train_data.samples}")
print(f"   Тестові зображення: {test_data.samples}")
print(f"   Класи: {list(train_data.class_indices.keys())}")


class_counts = {class_name: len(os.listdir(f'{base_dir}/train/{class_name}')) 
                for class_name in ['orchids', 'roses']}
print(f"Баланс класів: {class_counts}")

print("\nСтворення моделі CNN для класифікації рослин")

def create_plants_cnn_model():
    model = models.Sequential([
       
        layers.RandomFlip("horizontal_and_vertical"),
        layers.RandomRotation(0.2),
        layers.RandomZoom(0.2),
        layers.RandomContrast(0.2),
        
       
        layers.Conv2D(32, (5, 5), activation='relu', padding='same', 
                     input_shape=(IMG_SIZE[0], IMG_SIZE[1], 3)),
        layers.BatchNormalization(),
        layers.MaxPooling2D(2, 2),
        layers.Dropout(0.1),
        
       
        layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(2, 2),
        layers.Dropout(0.2),
        
       
        layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(2, 2),
        layers.Dropout(0.3),
        
        
        layers.Conv2D(256, (3, 3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(2, 2),
        layers.Dropout(0.4),
        
    
        layers.GlobalAveragePooling2D(),
        layers.Dense(512, activation='relu'),
        layers.BatchNormalization(),
        layers.Dropout(0.5),
        layers.Dense(256, activation='relu'),
        layers.BatchNormalization(),
        layers.Dropout(0.4),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(1, activation='sigmoid')
    ])
    
    return model


model = create_plants_cnn_model()


model.summary()

print("\nКомпіляція моделі для класифікації рослин")
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='binary_crossentropy',
    metrics=['accuracy', 'precision', 'recall']
)

print("Навчання моделі на зображеннях рослин:")


callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor='val_accuracy',
        patience=15,
        restore_best_weights=True,
        mode='max'
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.3,
        patience=8,
        min_lr=0.000001
    ),
    tf.keras.callbacks.ModelCheckpoint(
        'best_plants_model.keras',  
        monitor='val_accuracy',
        save_best_only=True,
        mode='max',
        verbose=1
    )
]


history = model.fit(
    train_data,
    epochs=50,
    validation_data=test_data,
    callbacks=callbacks,
    verbose=1
)

print("\nВізуалізація результатів навчання:")


plt.figure(figsize=(18, 6))

plt.subplot(1, 3, 1)
plt.plot(history.history['accuracy'], 'b-', label='Точність на тренуванні', linewidth=2)
plt.plot(history.history['val_accuracy'], 'r-', label='Точність на валідації', linewidth=2)
plt.title('Точність: Орхідеї проти троянд', fontweight='bold', fontsize=12)
plt.xlabel('Епоха')
plt.ylabel('Точність')
plt.legend()
plt.grid(True, alpha=0.3)


plt.subplot(1, 3, 2)
plt.plot(history.history['loss'], 'b-', label='Втрата на тренуванні', linewidth=2)
plt.plot(history.history['val_loss'], 'r-', label='Втрата на валідації', linewidth=2)
plt.title('Втрати: Орхідеї проти троянд', fontweight='bold', fontsize=12)
plt.xlabel('Епоха')
plt.ylabel('Втрати')
plt.legend()
plt.grid(True, alpha=0.3)


plt.subplot(1, 3, 3)
plt.plot(history.history['precision'], 'g-', label='Точність на тренуванні', linewidth=2, alpha=0.7)
plt.plot(history.history['recall'], 'y-', label='Повнота на тренуванні', linewidth=2, alpha=0.7)
if 'val_precision' in history.history:
    plt.plot(history.history['val_precision'], 'g--', label='Точність на валідації', linewidth=2)
    plt.plot(history.history['val_recall'], 'y--', label='Повнота на валідації', linewidth=2)
plt.title('Точність та Повнота', fontweight='bold', fontsize=12)
plt.xlabel('Епоха')
plt.ylabel('Оцінка')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("Детальна оцінка моделі")


try:
    best_model = tf.keras.models.load_model('best_plants_model.keras')
    print("Найкраща модель завантажена успішно\n")
except Exception as e:
    print(f"Найкраща модель не знайдена, використовуємо останню модель. Помилка: {e}")
    best_model = model


test_loss, test_accuracy, test_precision, test_recall = best_model.evaluate(test_data, verbose=0)
test_f1 = 2 * (test_precision * test_recall) / (test_precision + test_recall)

print(f"Результати класифікації рослин:")
print(f"Орхідеї проти троянд")
print(f"Точність: {test_accuracy:.4f} ({test_accuracy*100:.2f}%)")
print(f"Precision: {test_precision:.4f}")
print(f"Recall: {test_recall:.4f}")
print(f"F1-Score: {test_f1:.4f}")
print(f"Втрати: {test_loss:.4f}")


test_data.reset()
y_pred = (best_model.predict(test_data) > 0.5).astype("int32")
y_true = test_data.classes

print("\nКласифікації:")
print(classification_report(y_true, y_pred, 
                          target_names=['Орхідеї', 'Троянди']))


plt.figure(figsize=(8, 6))
cm = confusion_matrix(y_true, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='RdYlBu', 
            xticklabels=['Орхідеї', 'Троянди'], 
            yticklabels=['Орхідеї', 'Троянди'],
            annot_kws={"size": 14})
plt.title('Матриця плутанини\nКласифікація рослин', fontweight='bold', fontsize=14)
plt.xlabel('Передбачені мітки', fontsize=12)
plt.ylabel('Справжні мітки', fontsize=12)
plt.show()

print("Тестування на прикладах рослин")

def predict_plant_image(model, image_path, img_size=IMG_SIZE):
    img = tf.keras.preprocessing.image.load_img(image_path, target_size=img_size)
    img_array = tf.keras.preprocessing.image.img_to_array(img)
    img_array = tf.expand_dims(img_array, 0)
    img_array /= 255.0
    
    prediction = model.predict(img_array)[0][0]
    class_name = "Троянда" if prediction > 0.5 else "Орхідея"
    confidence = prediction if prediction > 0.5 else 1 - prediction
    
    plt.figure(figsize=(8, 8))
    plt.imshow(img)
    plt.title(f'Результат: {class_name}\nВпевненість: {confidence:.2%}', 
              fontweight='bold', fontsize=14, pad=20)
    plt.axis('off')
    plt.tight_layout()
    plt.show()
    
    return class_name, confidence

print("Демонстрація роботи моделі:")
try:
    demo_img = np.random.randint(100, 200, (224, 224, 3), dtype=np.uint8)
    center_x, center_y = 112, 112
    for angle in np.linspace(0, 2*np.pi, 8):
        x = int(center_x + 50 * np.cos(angle))
        y = int(center_y + 50 * np.sin(angle))
        for dx in range(-15, 16):
            for dy in range(-10, 11):
                if dx*dx/225 + dy*dy/100 <= 1:
                    nx, ny = x + dx, y + dy
                    if 0 <= nx < 224 and 0 <= ny < 224:
                        demo_img[ny, nx] = [238, 130, 238]  
    
    Image.fromarray(demo_img).save("demo_orchid.jpg")
    
    class_name, confidence = predict_plant_image(best_model, "demo_orchid.jpg")
    print(f"Результат: {class_name} - {confidence:.2%}")
    
except Exception as e:
    print(f"Помилка тестування: {e}")

print("\nЗбереження результатів:")


try:
    best_model.save('plants_classifier_final.keras')  
    print("Модель збережена як 'plants_classifier_final.keras'")
except Exception as e:
    print(f"Помилка збереження у сучасному форматі: {e}")
    try:
        model.save('plants_classifier_final.h5')
        print("Модель збережена як 'plants_classifier_final.h5' (legacy формат)")
    except Exception as e2:
        print(f"Помилка збереження: {e2}")


plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Точність на тренуванні')
plt.plot(history.history['val_accuracy'], label='Точність на валідації')
plt.title('Точність моделі: Орхідеї проти Троянд')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Втрата на тренуванні')
plt.plot(history.history['val_loss'], label='Втрата на валідації')
plt.title('Втрати моделі: Орхідеї проти Троянд')
plt.legend()

plt.tight_layout()
plt.savefig('plants_training_history.png', dpi=300, bbox_inches='tight')
print("Графіки навчання збережено")

print(f"\nМодель успішно навчена розрізняти орхідеї та троянди")
```
